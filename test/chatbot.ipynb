## Chatbot Sanity Check

This notebook is used to check the output from our quantized models to make sure they are reasonable. 

### 2-bit baseline

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Load model and tokenizer
model_name = "BrownianNotion/TinyLlama_v1.1_mix_wikitext_alpaca_2bit_BitDistiller_baseline"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto")

# Test prompt
prompt = "Once upon a time in a distant land,"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate text
outputs = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)

# Decode and print the result
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(result)
```

```python
# Test prompt
prompt = "The days of the week are"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate text
outputs = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)

# Decode and print the result
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(result)
```

```python
# Test prompt
prompt = "Explain to me which value is largest: a) 1 b) 0.5 c) -2 d) three"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate text
outputs = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)

# Decode and print the result
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(result)
```

### 1-bit model

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Load model and tokenizer
model_name = "fredericowieser/TinyLlama_v1.1_mix_wikitext_alpaca_1bit_BitDistiller_baseline"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Ensure only model weights are loaded, skipping unnecessary optimizer states
model = AutoModelForCausalLM.from_pretrained(
    model_name, 
    device_map="auto", 
    trust_remote_code=True, 
    low_cpu_mem_usage=True
)

# Test prompt
prompt = "Once upon a time in a distant land,"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate text
outputs = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)

# Decode and print the result
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(result)
```
